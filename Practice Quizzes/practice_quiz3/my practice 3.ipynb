1️⃣ Replace Values in a Column

Problem:
You have a dataset of students’ majors. Some entries are incorrect. Replace the wrong values with the correct ones in one .replace() call.

Task:
Replace 'Busniess' → 'Business' and 'Phsycology' → 'Psychology'.

```python
import pandas as pd

data = {
    'Student': ['Ava', 'Ben', 'Chloe', 'Dylan', 'Ella', 'Finn'],
    'Major': ['Busniess', 'Psychology', 'Busniess', 'Biology', 'Phsycology', 'Math']
}

df = pd.DataFrame(data)
df
```

```python
df['Major'] = df['Major'].replace({'Busniess': 'Business', 'Phsycology': 'Psychology'})
df
```

2️⃣ Subset → Group → Aggregate → Filter in One Line

Problem:
Using the dataset below, find the average score by subject for students who scored above 60, and keep only subjects where the average score ≥ 80.

```python
data = {
    'Student': ['Ava', 'Ben', 'Chloe', 'Dylan', 'Ella', 'Finn', 'Grace', 'Hank'],
    'Subject': ['Math', 'Math', 'Science', 'Science', 'Math', 'Science', 'Math', 'Science'],
    'Score': [92, 58, 84, 77, 88, 91, 79, 83]
}
df = pd.DataFrame(data)
```

```python
df[df['Score'] > 60].groupby('Subject')['Score'].mean().\
    reset_index().query('Score >= 80')
```

3️⃣ Recode Numeric Variable into Categories

Problem:
Given ages of employees, create a new column 'Age_Group' with:

"Young" for < 30

"Mid" for 30–50

"Senior" for > 50

```python
data = {'Name': ['Ava', 'Ben', 'Chloe', 'Dylan', 'Ella'],
        'Age': [25, 32, 47, 53, 29]}
df = pd.DataFrame(data)
```

```python
#solution
df['Age_Group'] = pd.cut(df['Age'],
                         bins=[0, 30, 50, 100],
                         labels=['Young', 'Mid', 'Senior'],
                         right=False)
df
```

4️⃣ Group, Aggregate, and Sort

Problem:
You have the following sales data. Group by 'Product', calculate total sales and average price, and sort by total sales (descending) — in one chained statement.

```python
data = {
    'Product': ['Pen', 'Pen', 'Notebook', 'Notebook', 'Pencil', 'Pencil'],
    'Units': [10, 15, 20, 25, 30, 10],
    'Price': [1.5, 1.4, 3.0, 3.2, 0.8, 1.0]
}
df = pd.DataFrame(data)
```

```python
#solution
df.groupby('Product', as_index=False).agg(
    total_sales=('Units', 'sum'),
    avg_price=('Price', 'mean')
).sort_values('total_sales', ascending=False)
```

5️⃣ Rename One Column

Problem:
Rename column 'Units' to 'Quantity' while keeping all other column names unchanged.

```python
df = pd.DataFrame({
    'Product': ['Pen', 'Pencil', 'Notebook'],
    'Units': [25, 40, 30],
    'Price': [1.5, 0.8, 3.0]
})
```

```python
#solution
df = df.rename(columns={'Units': 'Quantity'})
df
```

6️⃣ Merge Two DataFrames

Problem:
Merge two DataFrames on 'Product'. Add the 'Category' column from the second DataFrame into the first.

```python
sales = pd.DataFrame({
    'Product': ['Pen', 'Pencil', 'Notebook'],
    'Price': [1.5, 0.8, 3.0]
})

categories = pd.DataFrame({
    'Product': ['Pen', 'Pencil', 'Notebook'],
    'Category': ['Stationery', 'Stationery', 'Office Supplies']
})
```

```python
#solution
merged_df = sales.merge(categories, on='Product', how='left')
merged_df
```
