Example:

2. You are given two lists of numbers, x_vals and y_vals, and must create a Numpy array where the components are related to those in the lists (for example, you might be asked to create a Numpy array with entries $ z_k = (x_k)^2\sin(y_k)$. Then generate a histogram of the values in the Numpy array, with labeled axes and a title. You will also be asked to change the color of the bars in the histogram.

Example:

Give the following lists, create a Numpy array where each element is the sum of the corresponding elements in x_vals and y_vals.

Plot a histogram of the values in this array. Label the x-axis "Value" and y-axis "Frequency" and give the plot the title "Histogram of x + y".

Change the color of the bars to "skyblue".



```python
import numpy as np
import matplotlib.pyplot as plt

x_vals = [1, 2, 3, 4, 5]
y_vals = [10, 20, 30, 40, 50]

# Create Numpy array using the formula z_k = x_k^2 - 5*sin(y_k)
z = np.array([x**2 - 5*np.sin(y) for x, y in zip(x_vals, y_vals)])

# Plot histogram
plt.hist(z, color="skyblue")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of z_k = x_k^2 - 5*sin(y_k)")
plt.show()
```

3. Generate random samples from two specified distributions and plot their histograms, side by side in a single figure. Each sub-figure will have its own labeled axes and title.

Example:

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate samples
normal_samples = np.random.normal(0, 1, 100)
uniform_samples = np.random.uniform(0, 1, 100)

# Create a single figure with two subplots
plt.figure(figsize=(10, 4))

# First subplot: Normal distribution
plt.subplot(1, 2, 1)  # 1 row, 2 columns, first subplot
plt.hist(normal_samples, color="lightgreen")
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

# Second subplot: Uniform distribution
plt.subplot(1, 2, 2)  # 1 row, 2 columns, second subplot
plt.hist(uniform_samples, color="salmon")
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.tight_layout()  # Adjust spacing
plt.show()
```

5. Read in a .csv file to create a Pandas DataFrame. Loop over a list of variable/column names, printing the mean of each variable.

6. Create a list of .csv file names and an additional empty list. Then in a for loop, read in each file to make a DataFrame, compute the mean of one of the variables/features/columns, and store it as an element of the originally empty list. Then print the list. 

Example:

```python
import pandas as pd

files = [
    "practice_quiz_data/united_states_co2_emissions.csv",
    "practice_quiz_data/china_co2_emissions.csv",
    "practice_quiz_data/india_co2_emissions.csv"
]

mean_emissions = []

for file in files:
    df = pd.read_csv(file)
    mean_emissions.append(df["total_emissions"].mean())

print(mean_emissions)
```
